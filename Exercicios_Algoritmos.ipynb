```python
print("Classificação de idade")

idade = int(input("Insira sua idade: "))

if idade <= 12:
  print("Criança")

elif idade <= 17:
  print("Adolescente")

elif idade <= 59:
  print("Adulto")

else:
  print("Idoso")
```

```python
print("Classificação de triangulo")

lado1 = float(input("Insira o primeiro lado: "))
lado2 = float(input("Insira o segundo lado: "))
lado3 = float(input("Insira o terceiro lado: "))

if lado1 == lado2 == lado3:
  print("Equilátero")

elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
  print("Isósceles")

else:
  print("Escaleno")
```

```python
print("Opções de pagamento da loja")

valor_produto = float(input("Insira o valor do produto: "))

print("Escolha a forma de pagamento:")
print("Dinheiro/Pix")
print("Cartão de crédito")
print("Cartão de débito")

opcao = input("Opção: ").lower()

if opcao == "pix" or opcao == "dinheiro":
    print(f"Valor pix/dinheiro: {valor_produto}")

elif opcao == "credito":
    print(f"Valor crédito: {valor_produto}")

elif opcao == "debito":
    print(f"Valor débito: {valor_produto}")

else:
    print("Opção inválida.")
```

```python
print("=== DESCONTO DA VENDA ===")

valor = float(input("Digite o valor da compra: "))

if valor <= 100:
    desconto = 10

elif valor <= 500:
    desconto = 15

else:
    desconto = 30

valor_desconto = valor * desconto / 100

valor_final = valor - valor_desconto

print(f"Desconto aplicado: {desconto}%")
print(f"Valor do desconto: R$ {valor_desconto:.2f}")
print(f"Valor final: R$ {valor_final:.2f}")
```
