```python
print ("Multiplicação")

a = float (input ("Insira um número:"))
b = float (input("Insira um número:"))

multiplicacao = a * b

print ("Multiplicação:", multiplicacao)
```

```python
print("Dobro e Triplo")

a = float(input("Insira um número:"))

dobro = a * 2
triplo = a * 3

print("Dobro:", dobro)
print("Triplo:", triplo)
```

```python
print("Antecessor e Sucessor")

a = float(input("Insira um número:"))

antecessor = a - 1
sucessor = a + 1

print("Antecessor:", antecessor)
print("Sucessor:", sucessor)
```

```python
print("Perimetro do Retangulo")

base = float(input("Insira a base:"))
altura = float(input("Insira a altura:"))

perimetro = 2 * (base + altura)

print("Perimetro:", perimetro)
```

```python
print("Area do Retangulo")

base = float(input("Insira a base:"))
altura = float(input("Insira a altura:"))

area = base * altura
print("Area:", area)
```

```python
print("Negativo ou positivo")

numero = float(input("Insira um número"))

if numero > 0:
  print("\n O número é positivo!")

elif numero < 0:
  print("\n O número é negativo!")

else:
  print ("\n O número é zero!")



```

```python
print("Numero impar ou Par")

numero = float(input("Insira um número"))

if numero % 2 == 0:
 print("\n O número é par!")

else:
 print("\n O número é impar!")
```

```python
print("Maior entre dois números")

num1 = float(input("Insira o primeiro número:"))
num2 = float(input("Insira o segundo número:"))

if num1 > num2:
  print("\n O primeiro número é maior!")

else:
  print("\n O segundo número é maior!")
```

```python
print("Maior entre tres numeros")

a = float(input("Insira o primeiro número:"))
b = float(input("Insira o segundo número:"))
c = float(input("Insira o terceiro número:"))

maior = a

if b > maior:
  maior = b
if c > maior:
  maior = c

print ("Esse é o maior: ", maior)

```
